```python
import numpy as np
array = np.array([11, 22, 33, 44, 55])
print("Original array:", array)
print("Addition of 2:", array + 2)
print("Multiplication by 3:", array * 3)
print("Division by 2:", array / 2)
```

```python
arr = np.array([1, 2, 3, 6, 4, 5])
print("array:", arr)
print("Reversed:", arr[::-1])
```

```python
x = np.array([1,2,3,4,5,1,2,1,1,1])

values, counts = np.unique(x, return_counts=True)
most_frequent = values[np.argmax(counts)]
print("Most frequent value:", most_frequent)
print("Indices:", np.where(x == most_frequent)[0])
```

```python
y = np.array([1,1,1,2,3,4,2,4,3,3,])

values, counts = np.unique(y, return_counts=True)
most_frequent = values[np.argmax(counts)]
print("Most frequent value:", most_frequent)
print("Indices:", np.where(y == most_frequent)[0])
```

```python
arr = np.array([[10, 20, 30],[40, 50, 60],[70, 80, 90]])
print("Accessing 1st row, 2nd column:", arr[0, 1])
print("Accessing 3rd row, 1st column:", arr[2, 0])
```

```python
Vish = np.linspace(10, 100, 25)

print("Array:", Vish)
print("Dimensions:", Vish.ndim)
print("Shape:", Vish.shape)
print("Total elements:", Vish.size)
print("Data type:", Vish.dtype)
print("Total bytes:", Vish.nbytes)

Vish_transpose = Vish.reshape(25, 1)

print("Transpose using reshape() : ")
print(Vish_transpose)

print("Doable using T attribute?", Vish.T)
```

```python
ucs420_Vish = np.array([[10, 20, 30, 40],[50, 60, 70, 80],[90, 15, 20, 35]])

print("Original array:")
print(ucs420_Vish)

print("Mean:", np.mean(ucs420_Vish))
print("Median:", np.median(ucs420_Vish))
print("Maximum:", np.max(ucs420_Vish))
print("Minimum:", np.min(ucs420_Vish))
print("Unique elements:", np.unique(ucs420_Vish))
reshaped_ucs420_Vish = ucs420_Vish.reshape(4, 3)

print("Reshaped array:")
print(reshaped_ucs420_Vish)
resized_ucs420_Vish = np.resize(ucs420_Vish, (2, 3))
print("Resized array:")
print(resized_ucs420_Vish)
```
