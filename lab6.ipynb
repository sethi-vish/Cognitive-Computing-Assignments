```python
import numpy as np
```

```python
temp = np.array([25, 28, 31, 35, 38, 27, 33, 40])

temp_new = temp + 2
print("Temperature after adding 2:", temp_new)
fahrenheit = (9/5) * temp + 32
print("Temperature in Fahrenheit:", fahrenheit)
greater_than_32 = temp[temp > 32]
print("Readings greater than 32:", greater_than_32)
count = np.sum(temp > 32)
print("Number of readings above 32:", count)
print("Vectorization performs operations on entire arrays at once, while Boolean indexing filters/selects data without loops, making processing much faster and more efficient.")
```

```python
steps = np.array([
    [5000, 6200, 7100],
    [8000, 7500, 9000],
    [4500, 5100, 4800],
    [9000, 8500, 9500]
])

print("Total steps:", np.sum(steps))
print("Mean steps:", np.mean(steps))
print("Max:", np.max(steps))
print("Min:", np.min(steps))
print("Total steps for each day:", np.sum(steps, axis=0))
print("Total steps for each user:", np.sum(steps, axis=1))
max_position = np.unravel_index(np.argmax(steps), steps.shape)
print("Position of maximum steps:", max_position)
print("Maximum steps:", steps[max_position])
```

```python
original = np.array([1, 2, 3, 4, 5, 6])

subset = original[1:5]
subset[0] = 999

print("Original:", original)
print("Subset:", subset)

copied_array = original[1:5].copy()
copied_array[0] = 500

print("Original:", original)
print("Copied array:", copied_array)

matrix = np.arange(1, 13).reshape(3, 4)

print("3 x 4 matrix:")
print(matrix)
print("First row:", matrix[0])
print("Last row:", matrix[-1])
print("Second column:", matrix[:, 1])

print("Rows 1-2 and columns 2-3:")
print(matrix[1:3, 2:4])

flattened = matrix.flatten()
raveled = matrix.ravel()

print("Flatten:", flattened)
print("Ravel:", raveled)

raveled[0] = 1000

print("After modifying by ravel:")
print("Original matrix:")
print(matrix)
print("Ravel:", raveled)

flattened[1] = 2000

print("After modifying by flatten:")
print("Original matrix:")
print(matrix)
print("Flatten:", flattened)

print("Shape:", matrix.shape)
print("Dimensions:", matrix.ndim)
print("Size:", matrix.size)
print("Data type:", matrix.dtype)
```

```python
X = np.array([
    [6, 70, 3],
    [5, 50, 6],
    [8, 80, 2],
    [4, 30, 8]
])

y = np.array([40, 65, 30, 85])

print("Shape of X:", X.shape)
print("Dimensions of X:", X.ndim)

print("X transpose:")
print(X.T)
XTX = X.T @ X
print("X.T @ X:")
print(XTX)

XTX_inverse = np.linalg.inv(XTX)
print("Inverse of X.T @ X:")
print(XTX_inverse)

beta = XTX_inverse @ X.T @ y
print("Coefficients:", beta)

print("The coefficients represent the effect of sleep hours, activity level and stress level on the assistance score.")
new_user = np.array([5, 40, 7])
predicted_score = new_user @ beta
print("Predicted assistance score:", predicted_score)
```
