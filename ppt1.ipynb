### 1. What is the data type of the result of: 6 / 2?

```python
# The data type is float
print(type(6 / 2))
```

### 2. What is the type of the result of: 6 // 2? (Note the double slash //)

```python
# The data type is int (floor division)
print(type(6 // 2))
```

### 3. What is the type of the result of: "hello" == "world"?

```python
# The data type is bool (boolean)
print(type("hello" == "world"))
```

### 4. Write the code to convert this phone number 123-456-7890 to a string

```python
phone_str = str("123-456-7890")
print(type(phone_str), phone_str)
```

### 5. Write the code to convert this number representing financial value "1234.56" to a floating point number

```python
finance_val = float("1234.56")
print(type(finance_val), finance_val)
```

### 6. Write the code to convert the following number representing employeeid "1001" to an integer

```python
emp_id = int("1001")
print(type(emp_id), emp_id)
```

### 7. What is the result of int(3.99)

```python
# This evaluates to 3 (it truncates the decimal part, it does not round up)
print(int(3.99))
```
